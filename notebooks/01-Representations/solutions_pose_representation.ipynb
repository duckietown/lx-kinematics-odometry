# Moving between frames

```python
# Run this cell to initialize the problem
# Tip: be consistent with your units - meters and radians are great choices
# Fact: humans understand degrees better than radians. To avoid confusion, input and output degrees, but do calculations in radians

duckie_pos_g = np.array([2, 0.4])   # Position of Duckiebot in map/global frame (in meters)
duckie_or_g = 110                   # Orientation of Duckiebot in map/global frame (in degrees)
obstacle_dist_to_duckie = 0.3       # Obstacle distance to the Duckiebot
obstacle_angle = 50                 # Obstacle angle with respect to Duckiebot (in degrees)
```

```python
# Let's find the answer!
# Step 1. Convert degrees to radians
rot_in_rad_a = np.deg2rad(duckie_or_g) # (np.pi * duckie_or_g) / 180.
rot_in_rad_b = np.deg2rad(obstacle_angle) # (np.pi * obstacle_angle) / 180.

# Step 2. Transformation matrix from Duckiebot (a) to origin (o)

T_oa = np.array(([np.cos(rot_in_rad_a),-np.sin(rot_in_rad_a),duckie_pos_g[0]],
                 [np.sin(rot_in_rad_a),np.cos(rot_in_rad_a),duckie_pos_g[1]],
                 [0, 0, 1]))

# Step 3. Transformation matrix from obstacle (b) to Duckiebot (a)

dx = obstacle_dist_to_duckie * np.cos(rot_in_rad_b)
dy = obstacle_dist_to_duckie * np.sin(rot_in_rad_b)
T_ab =  np.array(([np.cos(rot_in_rad_b),-np.sin(rot_in_rad_b),dx],
                 [np.sin(rot_in_rad_b),np.cos(rot_in_rad_b),dy],
                 [0, 0, 1]))

# Step 4. Use the above to compute the transformation from obstacle (b) to origin (o)

T_ob = T_oa @ T_ab # T_ob = (T_oa)(T_ab)

# Step 5. Extract the information requested from the transformation matrix

obstacle_x_g = T_ob[0,2]
obstacle_y_g = T_ob[1,2]

# Write your answer (position of obstacle in global frame) here instead of "None, None" 
obstacle_pos_g = np.array([obstacle_x_g, obstacle_y_g])    
print(obstacle_pos_g)







```

```python
# Let's send the Duckiebot to the correct place!
# Convert degrees to radians
rot_in_rad_a = (np.pi * duckie_or_g) / 180.

# Write the transformation matrix from Duckiebot (a) to origin (o)
T_oa = np.array(([np.cos(rot_in_rad_a),-np.sin(rot_in_rad_a),duckie_pos_g[0]],
                 [np.sin(rot_in_rad_a),np.cos(rot_in_rad_a),duckie_pos_g[1]],
                 [0, 0, 1])) # Transformation matrix from Duckiebot (a) to origin (o)

# Write the transformation matrix from obstacle (b) to origin (o)             
T_ob = np.array(([1, 0, obstacle_pos_g[0]],
                 [0, 1, obstacle_pos_g[1]],
                 [0, 0, 1])) # Transformation matrix from obstacle (b) to origin (o)

# Leverage the known properties of transformation matrices and basics of linear algebra to find your answer
T_ab = np.linalg.inv(T_oa) @ T_ob


# Extract the obstacle position from the transformation matrix
obstacle_x_a = T_ab[0,2]
obstacle_y_a = T_ab[1,2]

# Place your answer here instead of None, None (position of obstacle in robot frame)
obstacle_pos_r = np.array([obstacle_x_a, obstacle_y_a])    
print(obstacle_pos_r)
```

```python
### Run this cell to initialize the problem
# Tip: remember to be consistent with your units

x_k_pos = np.array([3.0, 2.0])     # Position of Duckiebot in global frame at time k (meters)
x_k_heading = 60                   # Orientation of Duckiebot in global frame at time k (degrees)

u_k_pos = np.array([2.0, 1.0])     # Change in position of the robot between k and k+1 (in the robot frame, meters)
u_k_heading = -30                  # Change in orientation of the robot between k and k+1 (in the robot frame, degrees)

u_k1_pos = np.array([1.0, -0.3])   # Change in position of the robot between k+1 and k+2 (in the robot frame, meters)
u_k1_heading = 60                  # Change in orientation of the robot between k+1 and k+2 (in the robot frame, degrees)
```

```python
# Let's compute the pose of the Duckiebot!

# Convert degrees to radians
# Write the transformation matrix for the Duckiebot at time k
x_k_heading_rad = np.pi * x_k_heading / 180
T_x_k = np.array(([np.cos(x_k_heading_rad),-np.sin(x_k_heading_rad),x_k_pos[0]],
                 [np.sin(x_k_heading_rad),np.cos(x_k_heading_rad),x_k_pos[1]],
                 [0, 0, 1]))

# Write the transformation matrix that takes the Duckiebot to time k+1
u_k_heading_rad = np.pi * u_k_heading / 180
T_u_k = np.array(([np.cos(u_k_heading_rad),-np.sin(u_k_heading_rad),u_k_pos[0]],
                 [np.sin(u_k_heading_rad),np.cos(u_k_heading_rad),u_k_pos[1]],
                 [0, 0, 1]))

# Write the transformation matrix that takes the Duckiebot to time k+2
u_k1_heading_rad = np.pi * u_k1_heading / 180
T_u_k1 = np.array(([np.cos(u_k1_heading_rad),-np.sin(u_k1_heading_rad),u_k1_pos[0]],
                 [np.sin(u_k1_heading_rad),np.cos(u_k1_heading_rad),u_k1_pos[1]],
                 [0, 0, 1]))

# Compose the transformations
T_x_k2 = T_x_k @ T_u_k @ T_u_k1

# Extract the position and orientation of the Duckiebot at time k+2
x_k2_pos = np.array([T_x_k2[0,2], T_x_k2[1,2]])
x_k2_heading = np.arctan2(T_x_k2[1,0], T_x_k2[0,0]) * 180 / np.pi

print(f"The position of the Duckiebot at time k+2 is {x_k2_pos} with orientation {x_k2_heading}")
```
